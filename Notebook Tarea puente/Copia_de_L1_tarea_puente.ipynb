## Parte B · Dos maneras de cambiar de base

En la clase del miércoles 14 escribieron $M(n) = M(\lfloor n/10 \rfloor) + n \bmod 10$, con $M(0) = 0$: la suma de las cifras de $n$. Fíjense en qué hace esa recursión: **separa la última cifra con `mod` y le pasa el resto a sí misma con `div`**.

Cambiar de base es exactamente lo mismo, con dos diferencias: en vez de 10 se usa la base $b$, y en vez de **sumar** las cifras se **ponen en fila**. Los dígitos de $n$ en base $b$ son los dígitos de $n \operatorname{div} b$, seguidos del dígito $n \bmod b$.

Escriban `a_base(n, b)` de **dos maneras**:

- `a_base_ciclo(n, b)`: con un `while`. Es la máquina misteriosa del ejercicio 8 de *Falsos amigos*, con `b` en lugar de 2.
- `a_base_rec(n, b)`: recursiva. ¿Cuál es el caso base? (Cuidado con el falso amigo de la clase 14: la llamada tiene que achicarse y tiene que llegar al caso base.)

Y la inversa, `de_base(digitos, b)`: es su `leer_numero` de la sesión 1, con `b` en lugar de 2.

````python
def a_base_ciclo(n, b):
    """CONTRATO
    Recibe: un entero n >= 0 y una base b >= 2.
    Devuelve: la lista de dígitos de n en base b, el más significativo primero.
    a_base_ciclo(0, b) devuelve [0].
    """
    if n == 0:
        return [0]

    digitos = []
    while n > 0:
        digitos.append(n % b)
        n = n // b

    digitos.reverse()
    return digitos

def a_base_rec(n, b):
    """Mismo contrato, escrita de forma recursiva."""
    if n < b:
        return [n]
    return a_base_rec(n // b, b) + [n % b]

def de_base(digitos, b):
    """CONTRATO
    Recibe: una lista de dígitos en base b, el más significativo primero.
    Devuelve: el entero que representan.
    """
    numero = 0
    for d in digitos:
        numero = numero * b + d
    return numero
````

````python
# Pruebas
for f in (a_base_ciclo, a_base_rec):
    assert f(1974, 6) == [1, 3, 0, 5, 0], f.__name__
    assert f(13, 2) == [1, 1, 0, 1], f.__name__
    assert f(255, 16) == [15, 15], f.__name__
    assert f(0, 7) == [0], f.__name__
    assert f(5, 6) == [5], f.__name__
assert de_base([1, 3, 0, 5, 0], 6) == 1974
import random
for _ in range(1000):
    n, b = random.randint(0, 10**6), random.randint(2, 16)
    assert a_base_ciclo(n, b) == a_base_rec(n, b)
    assert de_base(a_base_rec(n, b), b) == n
print("✔ Las dos maneras coinciden en 1000 casos al azar, y de_base las deshace.")
````
